```python
import pandas as pd
import numpy as np
```

```python
scores = pd.read_csv("spreadspoke_scores.csv")
teams = pd.read_csv("nfl_teams.csv")

scores["team_home_id"] = scores["team_home"].map(teams.set_index("team_name")["team_id"])
scores["team_away_id"] = scores["team_away"].map(teams.set_index("team_name")["team_id"])
print(scores.head())

modern_scores = scores[(scores["schedule_season"] > 2005)]
```

```python
bears = modern_scores[(modern_scores['team_home'] == 'Chicago Bears') | (modern_scores['team_away'] == 'Chicago Bears')]
```

```python
seahawks_name = 'Seattle Seahawks'
seahawks = modern_scores[(modern_scores['team_home'] == seahawks_name) | (modern_scores['team_away'] == seahawks_name)]
print(seahawks.head())
```

```python
lions_vs_seahawks = seahawks[(seahawks["team_home"] == "Detroit Lions") | (seahawks["team_away"] == "Detroit Lions")]
print(lions_vs_seahawks.tail())
```

```python
#Find the biggest upset
scores["score_home_diff"] = scores["score_home"] - scores["score_away"]
scores["score_away_diff"] = scores["score_away"] - scores["score_home"]

scores["upset_score"] = np.where(scores['team_favorite_id'] == scores['team_home_id'], scores['score_away_diff'] - scores['spread_favorite'], scores['score_home_diff'] - scores['spread_favorite'] )
#print(scores[['team_home_id', 'team_away_id', 'team_favorite_id', 'score_home', 'score_away', 'spread_favorite', 'score_home_diff', 'score_away_diff', 'upset_score']])
upset_score_sorted = scores.sort_values(by='upset_score', ascending=False)

print(upset_score_sorted.head())
```
